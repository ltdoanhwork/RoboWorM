```python

import json
import os

def make_meta_txts(jsonl_path, dataset_dir):
    metas = os.path.join(dataset_dir, "metas")
    os.makedirs(metas, exist_ok=True)

    with open(jsonl_path, "r") as f:
        for line in f:
            ep = json.loads(line)
            idx = ep["episode_index"]
            tasks = ep["tasks"]

            out_path = os.path.join(metas, f"episode_{idx:06d}.txt")
            with open(out_path, "w") as out:
                out.write("\n".join(tasks))
```
